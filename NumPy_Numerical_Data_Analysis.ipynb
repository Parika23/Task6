## 1. Importing NumPy

NumPy is imported using the conventional alias `np`.


```python
import numpy as np

print("NumPy version:", np.__version__)
```

## 2. Creating 1D, 2D, and 3D Arrays

NumPy arrays can represent data in different dimensions:
- **1D:** a sequence of values
- **2D:** rows and columns, similar to a table
- **3D:** multiple 2D arrays stacked together


```python
# 1D array
array_1d = np.array([10, 20, 30, 40, 50])

# 2D array
array_2d = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

# 3D array
array_3d = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

print("1D array:")
print(array_1d)

print("\n2D array:")
print(array_2d)

print("\n3D array:")
print(array_3d)
```

## 3. Checking Dimensions, Shape, Size, and Data Type

The `ndim`, `shape`, `size`, and `dtype` attributes provide useful information about an array.


```python
for name, arr in [
    ("1D", array_1d),
    ("2D", array_2d),
    ("3D", array_3d)
]:
    print(f"{name} array:")
    print("  Dimensions (ndim):", arr.ndim)
    print("  Shape:", arr.shape)
    print("  Number of elements (size):", arr.size)
    print("  Data type (dtype):", arr.dtype)
```

## 4. Indexing and Slicing

NumPy uses zero-based indexing. Slicing allows a range of elements, rows, or columns to be selected.


```python
print("1D first element:", array_1d[0])
print("1D last element:", array_1d[-1])
print("1D elements from index 1 to 3:", array_1d[1:4])

print("\n2D element at row 1, column 2:", array_2d[1, 2])
print("2D first row:", array_2d[0, :])
print("2D second column:", array_2d[:, 1])
print("2D first two columns:")
print(array_2d[:, :2])

print("\n3D first 2D block:")
print(array_3d[0])
```

## 5. Reshaping Arrays

The `reshape()` function changes the shape of an array without changing its data. The total number of elements must remain the same.


```python
numbers = np.arange(1, 13)

reshaped_2d = numbers.reshape(3, 4)
reshaped_3d = numbers.reshape(2, 2, 3)

print("Original 1D array:")
print(numbers)

print("\nReshaped to 3 x 4:")
print(reshaped_2d)

print("\nReshaped to 2 x 2 x 3:")
print(reshaped_3d)
```

## 6. Generating Arrays with NumPy Functions

NumPy provides convenient functions for creating arrays:
- `zeros()` creates arrays filled with zeros.
- `ones()` creates arrays filled with ones.
- `arange()` generates values over a specified interval.
- `linspace()` generates evenly spaced values between two endpoints.


```python
zeros_array = np.zeros((2, 3))
ones_array = np.ones((2, 3))
arange_array = np.arange(0, 10, 2)
linspace_array = np.linspace(0, 1, 6)

print("zeros():")
print(zeros_array)

print("\nones():")
print(ones_array)

print("\narange(0, 10, 2):")
print(arange_array)

print("\nlinspace(0, 1, 6):")
print(linspace_array)
```

## 7. Mathematical Operations

NumPy supports element-wise mathematical operations directly on arrays. This makes numerical calculations concise and efficient.


```python
a = np.array([1, 2, 3, 4])
b = np.array([5, 6, 7, 8])

print("a + b:", a + b)
print("a - b:", a - b)
print("a * b:", a * b)
print("b / a:", b / a)
print("a squared:", a ** 2)
print("Square root of a:", np.sqrt(a))
```

## 8. Vectorized Operations

Vectorization allows an operation to be applied to an entire array without writing an explicit Python loop for each element.


```python
marks = np.array([55, 68, 72, 81, 90])

# Add 5 bonus marks to every student
adjusted_marks = marks + 5

# Convert marks to percentages out of 100
percentage = marks / 100 * 100

print("Original marks:", marks)
print("After adding 5 bonus marks:", adjusted_marks)
print("Percentage values:", percentage)
```

## 9. Boolean Masking

Boolean masking selects elements that satisfy a condition. It is commonly used for filtering numerical datasets.


```python
data = np.array([12, 25, 7, 30, 18, 42, 9, 35])

mask = data > 20
filtered_data = data[mask]

print("Original data:", data)
print("Boolean mask:", mask)
print("Values greater than 20:", filtered_data)

print("\nValues between 10 and 30:")
print(data[(data >= 10) & (data <= 30)])
```

## 10. Broadcasting

Broadcasting allows NumPy to perform operations on arrays with compatible shapes without manually replicating values.


```python
matrix = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

row_values = np.array([1, 2, 3])

result = matrix + row_values

print("Matrix:")
print(matrix)

print("\nValues broadcast across each row:")
print(row_values)

print("\nResult:")
print(result)
```

## 11. Aggregate Functions

Aggregate functions summarize numerical data. Common functions include `sum()`, `mean()`, `min()`, `max()`, and `median()`.


```python
scores = np.array([72, 85, 91, 68, 77, 95, 88])

print("Scores:", scores)
print("Sum:", np.sum(scores))
print("Mean:", np.mean(scores))
print("Minimum:", np.min(scores))
print("Maximum:", np.max(scores))
print("Median:", np.median(scores))
```

## 12. Aggregate Functions Along an Axis

For multidimensional arrays, aggregate functions can be applied across a particular axis.


```python
sales = np.array([
    [120, 150, 130],
    [200, 180, 210],
    [160, 170, 155]
])

print("Sales data:")
print(sales)

print("\nSum of each column:", np.sum(sales, axis=0))
print("Mean of each column:", np.mean(sales, axis=0))
print("Minimum of each column:", np.min(sales, axis=0))
print("Maximum of each column:", np.max(sales, axis=0))
print("Median of each column:", np.median(sales, axis=0))
```

## 13. Combined Numerical Analysis Example

The following example combines array creation, vectorized calculation, boolean masking, and aggregate functions to perform a small numerical analysis task.


```python
daily_sales = np.array([120, 145, 98, 175, 210, 160, 190])

# Apply a 10% increase using vectorization
projected_sales = daily_sales * 1.10

# Select days with sales above the original average
above_average = projected_sales > projected_sales.mean()
high_sales_days = projected_sales[above_average]

print("Original daily sales:", daily_sales)
print("Projected sales after 10% increase:", projected_sales)
print("Projected average sales:", projected_sales.mean())
print("Projected sales above average:", high_sales_days)
print("Highest projected sales:", projected_sales.max())
```
