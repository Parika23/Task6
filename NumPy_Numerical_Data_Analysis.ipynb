# NumPy for Numerical Data Analysis

This notebook demonstrates the fundamental NumPy operations used for numerical data analysis:

- Creating 1D, 2D, and 3D arrays
- Checking dimensions, shape, size, and data types
- Indexing and slicing
- Reshaping arrays
- Generating arrays with `zeros()`, `ones()`, `arange()`, and `linspace()`
- Mathematical and vectorized operations
- Boolean masking
- Broadcasting
- Aggregate functions such as `sum()`, `mean()`, `min()`, `max()`, and `median()`

Each section contains executable examples and their outputs.

## 1. Import NumPy

```python
import numpy as np

print("NumPy version:", np.__version__)
```

## 2. Creating 1D, 2D, and 3D Arrays

```python
# 1D array
array_1d = np.array([10, 20, 30, 40, 50])

# 2D array
array_2d = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

# 3D array
array_3d = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

print("1D Array:")
print(array_1d)

print("\n2D Array:")
print(array_2d)

print("\n3D Array:")
print(array_3d)
```

## 3. Checking Dimensions, Shape, Size, and Data Type

```python
print("1D dimensions:", array_1d.ndim)
print("1D shape:", array_1d.shape)
print("1D size:", array_1d.size)
print("1D data type:", array_1d.dtype)

print("\n2D dimensions:", array_2d.ndim)
print("2D shape:", array_2d.shape)
print("2D size:", array_2d.size)
print("2D data type:", array_2d.dtype)

print("\n3D dimensions:", array_3d.ndim)
print("3D shape:", array_3d.shape)
print("3D size:", array_3d.size)
print("3D data type:", array_3d.dtype)
```

## 4. Indexing and Slicing

```python
# 1D indexing and slicing
print("First element:", array_1d[0])
print("Last element:", array_1d[-1])
print("Elements from index 1 to 3:", array_1d[1:4])

# 2D indexing
print("\nElement at row 0, column 1:", array_2d[0, 1])
print("First row:", array_2d[0, :])
print("Second column:", array_2d[:, 1])

# 3D indexing
print("\nElement at [1, 0, 1]:", array_3d[1, 0, 1])
print("First 2D block:")
print(array_3d[0])
```

## 5. Reshaping Arrays

```python
original = np.arange(1, 13)
reshaped_2d = original.reshape(3, 4)
reshaped_3d = original.reshape(2, 2, 3)

print("Original array:")
print(original)

print("\nReshaped to 3 x 4:")
print(reshaped_2d)

print("\nReshaped to 2 x 2 x 3:")
print(reshaped_3d)

print("\nFlattened back to 1D:")
print(reshaped_2d.reshape(-1))
```

## 6. Generating Arrays with NumPy Functions

```python
zeros_array = np.zeros((2, 3))
ones_array = np.ones((2, 3))
arange_array = np.arange(0, 10, 2)
linspace_array = np.linspace(0, 1, 5)

print("zeros():")
print(zeros_array)

print("\nones():")
print(ones_array)

print("\narange():")
print(arange_array)

print("\nlinspace():")
print(linspace_array)
```

## 7. Mathematical Operations

```python
a = np.array([10, 20, 30, 40])
b = np.array([2, 4, 5, 8])

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Power:", a ** 2)
print("Square root:", np.sqrt(a))
print("Exponential:", np.exp(np.array([1, 2, 3])))
```

## 8. Vectorized Operations

```python
# NumPy performs the operation on every element without an explicit Python loop.
temperatures = np.array([20, 25, 30, 35, 40])

fahrenheit = temperatures * 9 / 5 + 32
adjusted = temperatures + 2

print("Temperatures in Celsius:", temperatures)
print("Temperatures in Fahrenheit:", fahrenheit)
print("After adding 2 degrees:", adjusted)
```

## 9. Boolean Masking

```python
scores = np.array([45, 72, 88, 56, 91, 67, 39, 84])

mask = scores >= 70

print("Scores:", scores)
print("Boolean mask:", mask)
print("Scores >= 70:", scores[mask])

# A more compact form
print("Scores between 50 and 90:", scores[(scores >= 50) & (scores <= 90)])
```

## 10. Broadcasting

```python
# A scalar is broadcast across every element.
data = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print("Original data:")
print(data)

print("\nAdd 5 to every element:")
print(data + 5)

# A 1D array is broadcast across the rows.
increments = np.array([1, 10, 100])

print("\nBroadcasted addition:")
print(data + increments)
```

## 11. Aggregate Functions

```python
values = np.array([10, 20, 30, 40, 50, 60])

print("Values:", values)
print("Sum:", np.sum(values))
print("Mean:", np.mean(values))
print("Minimum:", np.min(values))
print("Maximum:", np.max(values))
print("Median:", np.median(values))

# Aggregates along an axis
matrix = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print("\nMatrix:")
print(matrix)
print("Column sums:", np.sum(matrix, axis=0))
print("Row sums:", np.sum(matrix, axis=1))
print("Column means:", np.mean(matrix, axis=0))
print("Row means:", np.mean(matrix, axis=1))
```

## 12. Mini Numerical Data Analysis Example

The following example combines several NumPy concepts to analyze a small set of student marks.

```python
marks = np.array([
    [78, 85, 92],
    [65, 70, 75],
    [88, 90, 95],
    [55, 60, 68],
    [92, 94, 89]
])

subjects = ["Python", "NumPy", "Statistics"]

print("Marks:")
print(marks)

print("\nAverage mark per subject:")
print(dict(zip(subjects, np.mean(marks, axis=0))))

print("\nAverage mark per student:")
print(np.mean(marks, axis=1))

print("\nOverall average:", np.mean(marks))
print("Highest mark:", np.max(marks))
print("Lowest mark:", np.min(marks))
print("Median mark:", np.median(marks))

print("\nStudents with overall average >= 80:")
student_averages = np.mean(marks, axis=1)
print(marks[student_averages >= 80])
```

## 13. Summary

NumPy provides efficient tools for numerical data analysis. In this notebook, we used:

- **Arrays:** 1D, 2D, and 3D structures
- **Array properties:** `ndim`, `shape`, `size`, and `dtype`
- **Indexing/slicing:** Selecting individual elements and subsets
- **Reshaping:** Changing array dimensions without changing the data
- **Array generation:** `zeros()`, `ones()`, `arange()`, and `linspace()`
- **Mathematical operations:** Arithmetic and functions such as `sqrt()`
- **Vectorization:** Applying operations to entire arrays efficiently
- **Boolean masking:** Filtering values using conditions
- **Broadcasting:** Performing operations between compatible array shapes
- **Aggregation:** `sum()`, `mean()`, `min()`, `max()`, and `median()`

These operations form the foundation for numerical computing and data analysis with Python.
